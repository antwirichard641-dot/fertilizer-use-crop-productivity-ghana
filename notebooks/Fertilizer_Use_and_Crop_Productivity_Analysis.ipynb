Cell 8 — Corrected Completeness Check

```python
year_columns = [
    col for col in fertilizer_df.columns
    if col.startswith("Y") and len(col) == 5 and col[1:].isdigit()
]

fertilizer_indicators = ghana_fertilizer[
    ghana_fertilizer["Element"].isin([
        "Agricultural Use",
        "Use per area of cropland"
    ])
].copy()

missing_check = fertilizer_indicators[year_columns].isnull().sum(axis=1)

check_table = fertilizer_indicators[
    ["Item", "Element", "Unit"]
].copy()

check_table["Missing Years"] = missing_check.values
check_table["Available Years"] = len(year_columns) - missing_check.values

print("Number of actual year columns:", len(year_columns))
print("Year range:", year_columns[0], "to", year_columns[-1])

display(check_table)
```

Cell 9 — Reshape Fertilizer Data

```python
fertilizer_long = fertilizer_indicators.melt(
    id_vars=["Item", "Element", "Unit"],
    value_vars=year_columns,
    var_name="Year",
    value_name="Value"
)

fertilizer_long["Year"] = (
    fertilizer_long["Year"]
    .str.replace("Y", "", regex=False)
    .astype(int)
)

fertilizer_long = fertilizer_long.rename(
    columns={"Item": "Nutrient"}
)

print("Long-format fertilizer dataset shape:")
print(fertilizer_long.shape)

display(fertilizer_long.head())
```

Cell 11 — Plot Fertilizer Use Trends

```python
for element in ["Agricultural Use", "Use per area of cropland"]:

    plot_data = fertilizer_long[
        fertilizer_long["Element"] == element
    ]

    plt.figure(figsize=(10, 5))

    for nutrient in plot_data["Nutrient"].unique():
        nutrient_data = plot_data[
            plot_data["Nutrient"] == nutrient
        ]

        plt.plot(
            nutrient_data["Year"],
            nutrient_data["Value"],
            label=nutrient
        )

    unit = plot_data["Unit"].iloc[0]

    plt.title(f"Ghana Fertilizer {element} (1961–2024)")
    plt.xlabel("Year")
    plt.ylabel(f"{element} ({unit})")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

Cell 12 — Calculate Long-Term Fertilizer Change

```python
fertilizer_change = (
    fertilizer_long[fertilizer_long["Year"].isin([1961, 2024])]
    .pivot_table(
        index=["Nutrient", "Element"],
        columns="Year",
        values="Value"
    )
)

fertilizer_change["Absolute Change"] = (
    fertilizer_change[2024] - fertilizer_change[1961]
)

fertilizer_change["Percent Change"] = (
    (fertilizer_change[2024] - fertilizer_change[1961])
    / fertilizer_change[1961]
    * 100
)

display(fertilizer_change.round(2))
```

Next cell — Load the crop dataset

```python
# Load the clean crop dataset from Project 1
crop_df = pd.read_csv("ghana_crop_data_clean.csv")

print("Clean crop dataset loaded successfully.")
print("Shape:", crop_df.shape)
print("Columns:", crop_df.columns.tolist())
```

Next cell — Extract crop yield data

```python
# Keep only crop yield records needed for Project 2
crop_yield = crop_df[
    crop_df["Element"] == "Yield"
][["Item", "Year", "Value", "Unit"]].copy()

crop_yield = crop_yield.rename(
    columns={"Value": "Yield_kg_ha"}
)

print("Crop yield dataset prepared.")
print("Shape:", crop_yield.shape)
print("\nCrops:")
print(crop_yield["Item"].unique())

print("\nYear range:")
print(crop_yield["Year"].min(), "to", crop_yield["Year"].max())

print("\nMissing values:")
print(crop_yield.isna().sum())
```

Next step — Prepare the fertilizer data

```python
# Keep fertilizer use per area of cropland
fertilizer_area = fertilizer_long[
    fertilizer_long["Element"] == "Use per area of cropland"
][["Nutrient", "Year", "Value", "Unit"]].copy()

fertilizer_area = fertilizer_area.rename(
    columns={"Value": "Fertilizer_kg_ha"}
)

print("Fertilizer intensity dataset prepared.")
print("Shape:", fertilizer_area.shape)

print("\nNutrients:")
print(fertilizer_area["Nutrient"].unique())

print("\nYear range:")
print(fertilizer_area["Year"].min(), "to", fertilizer_area["Year"].max())

print("\nMissing values:")
print(fertilizer_area.isna().sum())

print("\nUnits:")
print(fertilizer_area["Unit"].value_counts())
```

Next step — Create a combined analysis dataset

```python
# Reshape fertilizer data so each nutrient becomes a separate column
fertilizer_wide = fertilizer_area.pivot(
    index="Year",
    columns="Nutrient",
    values="Fertilizer_kg_ha"
).reset_index()

# Rename nutrient columns for easier use
fertilizer_wide = fertilizer_wide.rename(columns={
    "Nutrient nitrogen N (total)": "Nitrogen_kg_ha",
    "Nutrient phosphate P2O5 (total)": "Phosphate_kg_ha",
    "Nutrient potash K2O (total)": "Potash_kg_ha"
})

print("Fertilizer dataset reshaped successfully.")
print("Shape:", fertilizer_wide.shape)

print("\nColumns:")
print(fertilizer_wide.columns.tolist())

print("\nFirst 5 rows:")
print(fertilizer_wide.head())

print("\nMissing values:")
print(fertilizer_wide.isna().sum())
```

Next cell — Merge the datasets

```python
# Merge crop yields with fertilizer intensity by year
analysis_df = crop_yield.merge(
    fertilizer_wide,
    on="Year",
    how="inner"
)

print("Datasets merged successfully.")
print("Shape:", analysis_df.shape)

print("\nColumns:")
print(analysis_df.columns.tolist())

print("\nYear range:")
print(analysis_df["Year"].min(), "to", analysis_df["Year"].max())

print("\nCrops:")
print(analysis_df["Item"].unique())

print("\nMissing values:")
print(analysis_df.isna().sum())

print("\nFirst 5 rows:")
print(analysis_df.head())
```

Next: visualize the relationship before calculating statistics

```python
# Plot crop yield against nitrogen fertilizer use per hectare

import matplotlib.pyplot as plt

crops = analysis_df["Item"].unique()

for crop in crops:
    plot_data = analysis_df[analysis_df["Item"] == crop]

    plt.figure(figsize=(7, 5))

    plt.scatter(
        plot_data["Nitrogen_kg_ha"],
        plot_data["Yield_kg_ha"]
    )

    plt.xlabel("Nitrogen fertilizer use (kg/ha)")
    plt.ylabel("Crop yield (kg/ha)")
    plt.title(f"{crop}: Yield vs Nitrogen Fertilizer Use")

    plt.tight_layout()
    plt.show()
```

Next cell — Calculate correlations

```python
# Calculate Pearson and Spearman correlations between nitrogen use and crop yield

from scipy.stats import pearsonr, spearmanr

correlation_results = []

for crop in analysis_df["Item"].unique():

    crop_data = analysis_df[analysis_df["Item"] == crop]

    pearson_r, pearson_p = pearsonr(
        crop_data["Nitrogen_kg_ha"],
        crop_data["Yield_kg_ha"]
    )

    spearman_rho, spearman_p = spearmanr(
        crop_data["Nitrogen_kg_ha"],
        crop_data["Yield_kg_ha"]
    )

    correlation_results.append({
        "Crop": crop,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p
    })

correlation_results_df = pd.DataFrame(correlation_results)

print("Nitrogen-yield correlation results:")
print(correlation_results_df.round(4))
```

Next step: control for the long-term trend

```python
# Calculate year-to-year changes in nitrogen use and crop yield

change_df = analysis_df.sort_values(
    ["Item", "Year"]
).copy()

change_df["Yield_change"] = (
    change_df.groupby("Item")["Yield_kg_ha"].diff()
)

change_df["Nitrogen_change"] = (
    change_df.groupby("Item")["Nitrogen_kg_ha"].diff()
)

# Remove the first year of each crop because it has no previous year
change_df = change_df.dropna(
    subset=["Yield_change", "Nitrogen_change"]
)

print("Year-to-year change dataset created.")
print("Shape:", change_df.shape)

print("\nYears:")
print(change_df["Year"].min(), "to", change_df["Year"].max())

print("\nMissing values:")
print(change_df[["Yield_change", "Nitrogen_change"]].isna().sum())

print("\nFirst 5 rows:")
print(
    change_df[
        ["Item", "Year", "Yield_change", "Nitrogen_change"]
    ].head()
)
```

Next cell — Correlate annual changes

```python
# Calculate correlations between year-to-year changes in nitrogen use and crop yield

change_correlation_results = []

for crop in change_df["Item"].unique():

    crop_data = change_df[change_df["Item"] == crop]

    pearson_r, pearson_p = pearsonr(
        crop_data["Nitrogen_change"],
        crop_data["Yield_change"]
    )

    spearman_rho, spearman_p = spearmanr(
        crop_data["Nitrogen_change"],
        crop_data["Yield_change"]
    )

    change_correlation_results.append({
        "Crop": crop,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_rho,
        "Spearman_p": spearman_p
    })

change_correlation_df = pd.DataFrame(change_correlation_results)

print("Year-to-year nitrogen-yield correlation results:")
print(change_correlation_df.round(4))
```

Next step: examine the other fertilizer nutrients

```python
# Calculate year-to-year correlations for all fertilizer nutrients and crop yields

nutrients = [
    "Nitrogen_kg_ha",
    "Phosphate_kg_ha",
    "Potash_kg_ha"
]

all_change_correlations = []

for crop in change_df["Item"].unique():

    crop_data = change_df[change_df["Item"] == crop]

    for nutrient in nutrients:

        nutrient_change = (
            crop_data[nutrient].diff()
        )

        # Remove the first observation created by the diff()
        valid_data = crop_data.copy()
        valid_data["Nutrient_change"] = nutrient_change
        valid_data = valid_data.dropna(
            subset=["Nutrient_change", "Yield_change"]
        )

        pearson_r, pearson_p = pearsonr(
            valid_data["Nutrient_change"],
            valid_data["Yield_change"]
        )

        spearman_rho, spearman_p = spearmanr(
            valid_data["Nutrient_change"],
            valid_data["Yield_change"]
        )

        all_change_correlations.append({
            "Crop": crop,
            "Nutrient": nutrient.replace("_kg_ha", ""),
            "Pearson_r": pearson_r,
            "Pearson_p": pearson_p,
            "Spearman_rho": spearman_rho,
            "Spearman_p": spearman_p
        })

all_change_correlations_df = pd.DataFrame(
    all_change_correlations
)

print("Year-to-year fertilizer-yield correlations:")
print(all_change_correlations_df.round(4))
```

Next cell — Corrected year-to-year analysis

```python
# Correct year-to-year correlation analysis for all fertilizer nutrients

nutrients = [
    "Nitrogen_kg_ha",
    "Phosphate_kg_ha",
    "Potash_kg_ha"
]

corrected_change_correlations = []

for crop in analysis_df["Item"].unique():

    crop_data = analysis_df[
        analysis_df["Item"] == crop
    ].sort_values("Year").copy()

    # Calculate annual changes once
    crop_data["Yield_change"] = (
        crop_data["Yield_kg_ha"].diff()
    )

    for nutrient in nutrients:

        crop_data["Nutrient_change"] = (
            crop_data[nutrient].diff()
        )

        valid_data = crop_data.dropna(
            subset=["Yield_change", "Nutrient_change"]
        )

        pearson_r, pearson_p = pearsonr(
            valid_data["Nutrient_change"],
            valid_data["Yield_change"]
        )

        spearman_rho, spearman_p = spearmanr(
            valid_data["Nutrient_change"],
            valid_data["Yield_change"]
        )

        corrected_change_correlations.append({
            "Crop": crop,
            "Nutrient": nutrient.replace("_kg_ha", ""),
            "Pearson_r": pearson_r,
            "Pearson_p": pearson_p,
            "Spearman_rho": spearman_rho,
            "Spearman_p": spearman_p,
            "N": len(valid_data)
        })

corrected_change_correlations_df = pd.DataFrame(
    corrected_change_correlations
)

print("Corrected year-to-year fertilizer-yield correlations:")
print(
    corrected_change_correlations_df.round(4)
)
```

Next step: make one final visualization

```python
# Visualize year-to-year changes in nitrogen use and crop yield

import matplotlib.pyplot as plt

crops = analysis_df["Item"].unique()

for crop in crops:

    crop_data = analysis_df[
        analysis_df["Item"] == crop
    ].sort_values("Year").copy()

    crop_data["Yield_change"] = (
        crop_data["Yield_kg_ha"].diff()
    )

    crop_data["Nitrogen_change"] = (
        crop_data["Nitrogen_kg_ha"].diff()
    )

    crop_data = crop_data.dropna(
        subset=["Yield_change", "Nitrogen_change"]
    )

    plt.figure(figsize=(7, 5))

    plt.scatter(
        crop_data["Nitrogen_change"],
        crop_data["Yield_change"]
    )

    plt.axhline(0, linewidth=0.8)
    plt.axvline(0, linewidth=0.8)

    plt.xlabel("Year-to-year change in nitrogen use (kg/ha)")
    plt.ylabel("Year-to-year change in yield (kg/ha)")
    plt.title(f"{crop}: Annual Changes in Yield vs Nitrogen Use")

    plt.tight_layout()
    plt.show()
```

Next step — simple regression on annual changes

```python
# Simple regression of year-to-year yield changes on fertilizer changes

from scipy.stats import linregress

regression_results = []

nutrients = [
    "Nitrogen_kg_ha",
    "Phosphate_kg_ha",
    "Potash_kg_ha"
]

for crop in analysis_df["Item"].unique():

    crop_data = analysis_df[
        analysis_df["Item"] == crop
    ].sort_values("Year").copy()

    crop_data["Yield_change"] = (
        crop_data["Yield_kg_ha"].diff()
    )

    for nutrient in nutrients:

        crop_data["Nutrient_change"] = (
            crop_data[nutrient].diff()
        )

        valid_data = crop_data.dropna(
            subset=["Yield_change", "Nutrient_change"]
        )

        result = linregress(
            valid_data["Nutrient_change"],
            valid_data["Yield_change"]
        )

        regression_results.append({
            "Crop": crop,
            "Nutrient": nutrient.replace("_kg_ha", ""),
            "Slope": result.slope,
            "R_squared": result.rvalue ** 2,
            "P_value": result.pvalue,
            "N": len(valid_data)
        })

regression_results_df = pd.DataFrame(
    regression_results
)

print("Year-to-year regression results:")
print(
    regression_results_df.round(4)
)
```
